# Student Grade Classification

```

# ==========================================
# PROJECT : Student Grade Classification
# CREATOR : Septian Yudhistira
# ==========================================

import time

student_score = 89

print("=" * 40)
print("    STUDENT GRADE CLASSIFICATION")
print("=" * 40)

time.sleep(1)
print(f"\n[STUDENT SCORE] : {student_score}")

print("\n[PROCESSING RESULT...]")
time.sleep(1)

if student_score >= 90:
    print("[RESULT] LULUS dengan Nilai A")
elif student_score >= 80:
    print("[RESULT] LULUS dengan Nilai B")
elif student_score >= 70:
    print("[RESULT] LULUS dengan Nilai C")
else:
    print("[RESULT] TIDAK LULUS / REMEDIAL")

time.sleep(1)
print("\n" + "=" * 40)
print("[STATUS] PROCESS COMPLETE")
print("=" * 40)
```
